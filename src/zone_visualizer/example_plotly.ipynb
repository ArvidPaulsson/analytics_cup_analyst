# Zone Visualizer - Plotly Example for Jupyter Notebooks

This notebook demonstrates how to use the Plotly-based zone visualizer in Jupyter notebooks.

The Plotly version provides:
- Interactive zooming and panning
- Hover tooltips with player information
- Smooth widget interactions
- Better performance in notebooks


```python
# Import required libraries
import sys
from pathlib import Path

# Add the analytics_cup_analyst directory to the path
# Notebook is in: analytics_cup_analyst/src/zone_visualizer/
# So we go up 2 levels to get to analytics_cup_analyst/
project_root = Path().resolve().parent.parent
sys.path.insert(0, str(project_root))

from src.zone_visualizer import ZoneLayer, ZoneConfig
```

## 1. Load Match Data

Initialize the zone layer with a match ID. The data will be automatically loaded.


```python
# Match ID from the dataset
match_id = 1886347

# Create zone configuration
zone_config = ZoneConfig(
    zone_0_radius=5.0,  # Zone 0 is just the ball carrier
    zone_1_radius=15.0,  # Support zone radius (meters)
    zone_1_angle=60.0,  # Angular cone for Zone 1 (±60 degrees)
    zone_1_orientation="toward_goal",  # Fallback orientation (ball movement direction takes priority)
    zone_2_radius=None,  # None means all remaining players
)

print(f"Loading match {match_id}...")

# Initialize zone layer
zone_layer = ZoneLayer(
    match_id=match_id,
    zone_config=zone_config,
    use_kloppy=True,  # Use kloppy if available
)

print("Data loaded successfully!")
print(f"Total frames: {len(zone_layer.tracking_data['frame'].unique())}")
```

## 2. Static Frame Visualization

Visualize a single frame using Plotly. The figure is returned and can be displayed or further customized.


```python
# Visualize a specific frame
example_frame = 100  # Choose a frame number

fig = zone_layer.visualize_frame_plotly(
    frame=example_frame,
    show_overlays=True,
    show_labels=True,
    show_zone_boundaries=True,
)

# Display the figure
fig.show()
```

## 3. Interactive Viewer

Launch the full interactive viewer with widgets for frame navigation and overlay toggles.

**Note**: Make sure you have `ipywidgets` installed: `pip install ipywidgets`


```python
# Launch interactive viewer with Plotly
zone_layer.launch_interactive(use_plotly=True)
```
